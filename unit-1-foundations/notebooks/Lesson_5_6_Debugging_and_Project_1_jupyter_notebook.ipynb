## Overview - Debugging Basics

In Python, data types determine how values behave and interact with each other. Some data types can be combined (or “operated on”) together, while others cannot. When combining data types, it’s important to understand their compatibility and how to handle mismatches.


### Key Points

1. Same Data Types Can Combine Easily:

Numbers: Integers (int) and floating-point numbers (float) can be added, subtracted, multiplied, and divided.
Strings: Strings can be combined (or concatenated) using the + operator.

2. Different Data Types Require Conversion:

Some combinations of data types are not allowed directly, but Python provides ways to convert between types.

3. Unsupported Combinations:

Python will raise a TypeError if you try to combine incompatible data types without converting them.

 **Examples**

Combining Numbers

* Allowed: Integer and float can combine directly.

````python
x = 5    # int
y = 2.5  # float
result = x + y  # Combines as a float
print(result)  # Output: 7.5
````

* Not Allowed: Combining a number with a string without conversion.

````python
x = 5    # int
y = "2"  # str
result = x + y  # TypeError: unsupported operand type(s)
````

## String Concatenation

* Allowed: Strings can be combined using the + operator.

````python
greeting = "Hello"
name = "Alice"
message = greeting + " " + name
print(message)  # Output: Hello Alice
````

* Not Allowed: Combining strings with numbers directly.

````python
age = 25
message = "I am " + age  # TypeError: can only concatenate str (not "int") to str
````

* Solution: Convert the number to a string using str().

````python
age = 25
message = "I am " + str(age)
print(message)  # Output: I am 25
````

Notice that wrapping the 25 in str() makes it a string data type, you can also do that by storing 25 as "25". Having multiple ways to do the same thing can be helpful because sometimes data is input as one data type and our program needs to change it to another.

##Type Conversion

You can explicitly convert one type to another to enable combinations:

* int() to convert to an integer.
* float() to convert to a float.
* str() to convert to a string.

**Example:**

````python
x = "10"  # str
y = 5     # int
result = int(x) + y  # Convert x to int before adding
print(result)  # Output: 15
````

**Unsupported Combinations**

Some data types, like lists and strings, cannot be directly combined.

````python
my_list = [1, 2, 3]
my_string = "hello"
result = my_list + my_string  # TypeError: can only concatenate list (not "str") to list
````

* Solution: Convert one of the types to make it compatible.

````python
my_list = [1, 2, 3]
my_string = "hello"
result = my_list + list(my_string)
print(result)  # Output: [1, 2, 3, 'h', 'e', 'l', 'l', 'o']
````

## Takeaways

1. **Be Mindful of Data Types:** Python is strict about combining incompatible types.

2. **Use Type Conversion**: When combining different types, convert them explicitly.

3. **Error Messages Are Your Friend:** If you see a TypeError, check your data types and consider conversion. If your errors are confusing, try throwing them into a google search. Most programmers have dealt with similar issues and they can help you figure out the issue. Let's go through the different types of errors you can expect to see in programming.



##Types of Errors



1. **Syntax Errors** – These occur when the code does not follow Python’s rules (grammar mistakes).

**Example:**

```
print("Hello, world!  

Error Message: SyntaxError: EOL while scanning string literal

Fix: Make sure all quotes, parentheses, and colons are correctly placed.

print("Hello, world!")  
```


2. **Runtime Errors** (Exceptions) – These occur while the program is running and cause it to stop.

**Example:**
```
num = 5 / 0  

Error Message: ZeroDivisionError: division by zero

Fix: Avoid dividing by zero by checking input values.
```


3. **Logical Errors** – These don’t produce error messages, but the program gives incorrect results.

**Example (Incorrect Calculation):**



````python
total = 5 * 2 + 3  # Intended: (5 * (2 + 3))
print(total)  # Output: 13 (wrong answer)
# Fix: Use parentheses to clarify order of operations.

total = 5 * (2 + 3)

print(total)  # Output: 25 (correct answer)
````

###How to Fix Errors in Your Code



1. Read the Error Message Carefully

Python provides useful messages when errors occur. Pay attention to:
The error type (e.g., SyntaxError, TypeError, IndexError).
The line number where the error occurred.


2. Check for Typos

Python is case-sensitive (Print() is incorrect; print() is correct).
Variable names must be spelled exactly as defined.


3. Use Print Statements for Debugging

If your program runs but gives the wrong output, use print() to check variable values.

**Example Code**


````python

x = 10

y = 5

print("Before Calculation: x =", x, "y =", y)

total = x - y  # Meant to add x + y

print("Total:", total)
````




4. Test Your Code in Small Parts

Instead of running the entire program at once, test smaller sections of code.



5. Use Online Resources

If you’re stuck, try searching for the error message online—chances are, someone else has had the same problem!



6. Ask for Help

Explaining your error to someone else often helps you find the mistake.

**Final Tip: Learn from Your Errors!**

Making mistakes is part of the learning process. The more errors you encounter and fix, the better you’ll become at debugging your code.

```

```

This splits the text at spaces and returns a list of words.



####3. Converting Text to Lowercase

Since Python is case-sensitive, "Hello" and "hello" are treated as different words. To ensure consistency, you should convert text to lowercase before analyzing it.
```
text = text.lower()
```
Now "The" and "the" will be treated as the same word.

####4. Removing Punctuation

Punctuation marks don’t add meaning in most text analysis tasks, so they should be removed.import string  
```
text = text.translate(str.maketrans("", "", string.punctuation))
```
• This removes commas, periods, question marks, etc.

• The string.punctuation constant includes all common punctuation marks.



####5. Counting Word Frequency

To find the most common words, you can use a Python dictionary to count how often each word appears.word_counts = {}

##### Create an empty dictionary  
```
for word in words:

    word_counts[word] = word_counts.get(word, 0) + 1  

print(word_counts)
```
This loop counts the occurrences of each word in the text.

####6. Finding the Most Common Words

To see which words appear most frequently, sort the dictionary by value.
```
sorted_words = sorted(word_counts.items(), key=lambda x: x[1], reverse=True)

print("Most common words:", sorted_words[:10])  # Print the 10 most common words
```
This sorts words by frequency from highest to lowest.
