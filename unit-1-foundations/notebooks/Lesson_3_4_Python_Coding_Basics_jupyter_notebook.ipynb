## Conditional Statements

Conditional statements are referred to as flow control statements because they affect the order in which code is executed and which code is executed based on the conditions you set. Flow control statements enable programs to make decisions, repeat actions, or skip sections of code based on conditions.

* These allow your code to make decisions based on conditions:

```
age = 16
if age >= 18:
    print("You can vote!")
else:
    print("You’re too young to vote.")
```

Notice you set the condition for the code to check, in this example, if the numerical value of age is greater than or equal to 18. If that statement is true then we go to the indented code, else, we go to the other code.

* Python uses indentation (spaces at the beginning of a line) to show blocks of code.

* These are not the only conditional statements and we will see the other phrases in the future.

````python
age = 16
if age >= 18:
    print("You can vote!")
else:
    print("You’re too young to vote.")
````

## Loops

Loops are another type of flow control statement. Loops allow you to repeat an action a certain amount of times depending on whether it is a 'for' look or a 'while' loop.

* For Loops: Repeat a task for a specific number of times:

```
for i in range(5):
    print("This is loop number", i)
```

Notice the structure of the text. The syntax of the 'for' loop confuses many students. A 'for' loop in Python is used to iterate over a variety of datatypes which generally contain a sequence of letters or numbers (such as a list, tuple, dictionary, string, or range) and execute a block of code multiple times. For each value in the sequence.

````python
for i in range(5):
    print("This is loop number", i)
````

###Basic Syntax of a for Loop

```
for variable in sequence:

    # Code to execute in each iteration
```

We've already looked at variables, in the 'for' loop it can be anything. By tradition, many coders just put i, sometimes x, but this could be a name or any other string that represents a variable.  We've already mentioned that any built-in function has to be indented to say that the entire block of text is one function. Which leaves the Sequence:

* sequence: The collection of elements (e.g., list, string, range) that the loop iterates through. This could be a list of numbers or letters.

##While Loops:

A while loop in Python executes a block of code repeatedly as long as a given condition remains True. Unlike a for loop, which iterates over a sequence, a while loop continues running until a condition is no longer met.


###Example 1: While Loop

```
count = 0
while count < 5:
    print("Count is", count)
    count += 1
Output:

Count is: 0  
Count is: 1  
Count is: 2  
Count is: 3  
Count is: 4  
 ```

````python
count = 0
while count < 5:
    print("Count is", count)
    count += 1
````

In this example, we say the count is equal to zero, and while it is less than 5, we print the count, and the number that count is, and then that symbol += adds 1 to count, so after the first iteration, count goes from 0 to 1. This will continue until count = 5, at which point the loop will end.

### Example 2: Using while for User Input

A while loop is useful when you don’t know how many times a loop should run beforehand.

```
password = ""  

while password != "secret":  
    password = input("Enter the password: ")  

print("Access granted!")  
```

If you read through this, notice that the password is initially defined as an empty string. Until you put the correct password in(secret), this loop will repeat. Notice the use of != to mean not equal to. Once the condition is met, the loop ends.

````python
password = ""

while password != "secret":
    password = input("Enter the password: ")

print("Access granted!")
````

### Example 3: Avoiding Infinite Loops

Be careful! If the condition never becomes False, the loop runs forever. There are certain system shortcuts to end these loops, but these are based on the computer you have. It's good to look these shortcuts up.

```
x = 10  

while x > 0:  
    print("Counting down:", x)  
    x -= 1  # Decreasing x each time to ensure the loop stops

```
 Without x -= 1 (which subtracts 1 from x each iteration), this loop will continue forever.


````python
x = 10

while x > 0:
    print("Counting down:", x)
    x -= 1  # Decreasing x each time to ensure the loop stops
````

###Example 4: Using break to Stop a While Loop

You can manually exit a while loop using break.

```
number = 1  

while number < 10:  
    print(number)  
    if number == 5:  
        print("Stopping the loop.")  
        break  # Stops the loop  
    number += 1  
 Output:

1  
2  
3  
4  
5  
Stopping the loop.  
 ```

**Explanation:**

• The loop stops early when the number reaches 5.

````python
number = 1

while number < 10:
    print(number)
    if number == 5:
        print("Stopping the loop.")
        break  # Stops the loop
    number += 1
````

### Example 5: Using continue to Skip an Iteration

The continue statement skips the rest of the loop and moves to the next iteration.

```
x = 0  

while x < 5:  
    x += 1  
    if x == 3:  
        continue  # Skips printing 3  
    print(x)  
Output:

1  
2  
4  
5  
 ```

**Explanation:**

* When x == 3, continue skips that iteration. The use of == makes this a truth statement, to say that it is equal, true or false

````python
x = 0

while x < 5:
    x += 1
    if x == 3:
        continue  # Skips printing 3
    print(x)
````
