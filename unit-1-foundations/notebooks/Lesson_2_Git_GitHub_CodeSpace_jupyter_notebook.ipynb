## Key Takeaways



1.   Indented Code Blocks

Python uses indentation to define code structure. Certain statements cause these blocks to occur automatically.

2.   The Colon (:) Trigger






Generally, any line of code that ends in a colon creates a new block of code on the next line.

3.   Control Flow Statements

Statements like "for" tell Python which code to execute first, allowing it to move beyond simple top-to-bottom execution.

While code normally executes top-to-bottom, Control Flow Statements allow us to skip or repeat sections based on logic.



### What is a for loop?

A for loop is a repeated instruction that tells the computer to do the same thing repeatedly, with slight variations each time.


### The range() function

range() is a built-in function that lists numbers starting at 0 up until the number of your choosing.

This allows us to display lists or change sentences dynamically.
