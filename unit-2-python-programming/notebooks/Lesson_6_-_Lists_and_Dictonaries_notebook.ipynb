###What is a List?

A list is a collection of ordered items in Python. Lists can store multiple values, including numbers, strings, and even other lists.



###Creating a List

To create a list, use square brackets [ ] and separate the items with commas.
```
Example Code:

# A list of numbers
numbers = [1, 2, 3, 4, 5]

# A list of strings
fruits = ["apple", "banana", "cherry"]

# A mixed list
mixed_list = [10, "hello", 3.5, True]

print(numbers)
print(fruits)
print(mixed_list)
Expected Output:

[1, 2, 3, 4, 5]
['apple', 'banana', 'cherry']
[10, 'hello', 3.5, True]
```
## Accessing and Modifying Lists
###Accessing Elements in a List

Each item in a list has an index starting from 0. You can access individual items using their index.
```
Example Code:

fruits = ["apple", "banana", "cherry"]

print(fruits[0])  # First item
print(fruits[1])  # Second item
print(fruits[-1]) # Last item
Expected Output:

apple
banana
cherry
Tip: Using -1 lets you access the last item in the list, -2 the second-last, and so on.
```


###Modifying Lists

You can change, add, and remove items in a list.
```
Example Code:

fruits = ["apple", "banana", "cherry"]

fruits[1] = "blueberry"  # Change "banana" to "blueberry"
fruits.append("orange")  # Add a new item
fruits.remove("apple")   # Remove an item

print(fruits)
Expected Output:

['blueberry', 'cherry', 'orange']
```

## Looping Through Lists
###Looping Through a List

You can use a for loop to go through each item in a list.
```
Example Code:

fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
Expected Output:

apple
banana
cherry
 ```
## Dictionaries
###What is a Dictionary?

A dictionary is a collection of key-value pairs. Unlike lists, where items are accessed by their index, dictionaries use keys to store and retrieve values.

###Creating a Dictionary

To create a dictionary, use curly braces { } with keys and values separated by colons :.
```
Example Code:

# A dictionary storing student information
student = {
    "name": "Alice",
    "age": 17,
    "grade": "A"
}

print(student)
Expected Output:

{'name': 'Alice', 'age': 17, 'grade': 'A'}

 ```
## Accessing and Modifying Dictionaries
###Accessing Dictionary Values

To access a value, use its key inside square brackets [ ] or the .get() method.
```
Example Code:

print(student["name"])   # Access using key
print(student.get("age"))  # Access using .get()
Expected Output:

Alice
17

```
###Modifying Dictionaries

You can add, change, and remove key-value pairs in a dictionary.

 ```

Example Code:

# Adding a new key-value pair
student["school"] = "High School"

# Changing a value
student["grade"] = "A+"

# Removing a key-value pair
del student["age"]

print(student)
Expected Output:

{'name': 'Alice', 'grade': 'A+', 'school': 'High School'}

 ```
## Looping Through Dictionaries
###Looping Through a Dictionary

You can loop through keys, values, or both in a dictionary.
```
for key, value in student.items():
    print(f"{key}: {value}")
Example Code:

for key, value in student.items():
    print(f"{key}: {value}")
Expected Output:

name: Alice
grade: A+
school: High School

```
