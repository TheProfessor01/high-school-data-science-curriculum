##Conditional Statements

###What Are Conditional Statements?

Conditional statements allow a program to make decisions based on certain conditions. In Python, we use if, elif (else if), and else statements to execute specific blocks of code depending on whether a condition is True or False.



**Why Are Conditional Statements Important?**

Conditional statements are used everywhere in programming. They allow computers to:

* Respond to user input (e.g., checking if a password is correct).
* Make decisions in games (e.g., if a player’s score reaches a certain value, they win).
* Filter and analyze data (e.g., only selecting numbers greater than 10 in a dataset).


**Basic Structure of an If Statement**

The if statement checks a condition. If the condition is True, the code inside the block runs. Otherwise, it is skipped.
```
age = 18  
if age >= 18:  
    print("You are allowed to vote.")  
```
**Explanation:** Since age is 18, which is greater than or equal to 18, the message “You are allowed to vote.” is printed.



###Using Else: What If the Condition Is False?

If the condition in an if statement is False, the code in an else block will run instead.

```
age = 16  
if age >= 18:  
    print("You are allowed to vote.")  
else:  
    print("You are too young to vote.")  
```
**Explanation:** Since age is 16, which is less than 18, the program prints “You are too young to vote.”


###Using Elif: Checking Multiple Conditions

Sometimes, we need to check multiple conditions. We can use elif (short for “else if”).
```
grade = 85  

if grade >= 90:  
    print("You got an A!")  
elif grade >= 80:  
    print("You got a B.")  
elif grade >= 70:  
    print("You got a C.")  
else:  
    print("You need to improve.")  
```

**Explanation:** Since grade is 85, the first condition is False, but the second (grade >= 80) is True, so the output is “You got a B.”



###Using Conditions with Comparison and Logical Operators

You can combine conditions using comparison operators (<, >, ==, !=, >=, <=) and logical operators (and, or, not).
```
temperature = 75  
is_raining = False  

if temperature > 70 and not is_raining:  
    print("It's a nice day to go outside!")  
else:  
    print("Better stay inside.")  
```
**Explanation:** The condition checks if temperature > 70 and it is not raining. Since both are True, the message “It’s a nice day to go outside!” is printed.



###Indentation: Why It Matters in Python



Python uses indentation (spaces or tabs) to define blocks of code inside an if, elif, or else statement. If the indentation is incorrect, you will get an error!

 Incorrect Example:
```
if x > 10:  
print("x is greater than 10")  # This will cause an IndentationError!  
```
```
Correct Example:

if x > 10:  
    print("x is greater than 10")  # Proper indentation!  
```

**Summary: Key Takeaways**

* If statements run code only when a condition is True.
* Else statements run when the if condition is False.
* Elif statements check multiple conditions in order.
* Comparison and logical operators allow us to create more complex conditions.
* Proper indentation is required in Python for conditional statements.

## Loops (For and While)

**What are Loops?**

Loops allow us to repeat a block of code multiple times without having to write it repeatedly. Loops are a key tool in data science because they allow us to process large datasets one element at a time. In Python, there are two main types of loops:


1. For Loops – Used when we know how many times we want to repeat an action.

2. While Loops – Used when we want to repeat an action until a certain condition is met.



###For Loops

A for loop is useful when we need to iterate over a sequence of items (like a list, string, or range of numbers).



####Basic Syntax:

for variable in sequence:
    # Code to repeat
• The loop runs once for each item in the sequence.

• The variable takes on the value of each item in the sequence one by one.



**Example 1: Looping Through a List**
```
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
Output:

apple  
banana  
cherry  
 ```

**Example 2: Using a Range**

We can use the range() function to loop a specific number of times.
```
for i in range(5):
    print(i)
Output:

0  
1  
2  
3  
4  
```

Remember: the built-in range function lists numbers from 0 up to the number you chose but not including it. There are ways you can vary the range function but you'd have to read the range documentation (Link)



###While Loops

A while loop runs as long as a certain condition remains True.

####Basic Syntax:

while condition:
    # Code to repeat
• The loop continues until the condition becomes False.



**Example 1: Counting Up to 5**
```
count = 1
while count <= 5:
    print(count)
    count += 1  # Increases count by 1 each time

Output:

1  
2  
3  
4  
5  
 ```

**Example 2: Asking for User Input Until They Enter “stop”**
```
user_input = ""
while user_input.lower() != "stop":
    user_input = input("Type something (or 'stop' to quit): ")
```

## Loop Control (break and continue)
###Break and Continue in Loops

 Python has two important loop control statements:

* break – Stops the loop immediately.

* continue – Skips the current iteration and moves to the next one.



**Example: Using break to Stop a Loop Early**
```
for number in range(10):
    if number == 5:
        break  # Stops the loop when number is 5
    print(number)

Output:

0  
1  
2  
3  
4  
 ```

**Example: Using continue to Skip an Iteration**
```
for number in range(5):
    if number == 2:
        continue  # Skips printing 2
    print(number)

Output:

0  
1  
3  
4  
```

###Common Mistakes and Debugging Tips

1. Infinite While Loops

If you forget to update the condition, a while loop can run forever!

  * Example of an infinite loop:
```
count = 1
while count > 0:  # This condition is always True
    print(count)
```
  * To fix it, make sure the condition eventually becomes False.



2. Off-by-One Errors in For Loops

  * Remember that range(n) stops before n.

  * If you want to loop exactly n times, use range(n), not range(n+1).



3. Forgetting to Use break Properly

  * If you need to stop a loop early, ensure break is inside an if statement.



###Conclusion

* For loops are great for iterating over sequences or repeating code a set number of times.

* While loops are best when looping until a condition is met.

* Break stops a loop early, and continue skips an iteration.
