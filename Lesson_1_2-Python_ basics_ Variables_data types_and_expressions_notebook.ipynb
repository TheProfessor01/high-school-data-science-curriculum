##Instruction - Methods

In your most recent activity, you used a method called .lower() to change the text string to lowercase. Methods are like functions (Ex: print(), input()) as they are built into the Python language and accomplish tasks for us.

A method in Python is a function that belongs to an object and is used to perform specific actions on that object. Think of a method as a built-in tool that objects (like strings, lists, or numbers) can use to do something useful.


**How is a Method Different from a Function?**

A function is a standalone block of code that performs a task. Example:
```
print("Hello!")  # print() is a function
```
A method is a function that belongs to an object and is used with a dot (.) notation. Example:
```
text = "hello"
print(text.upper())  # upper() is a method that belongs to string objects
 ```

###Using Methods in Python

Let’s look at some examples of methods for different types of objects.



####String Methods

Python has built-in methods that allow you to modify text (strings).
```
word = "python"
print(word.upper())  # Converts to uppercase: "PYTHON"
print(word.lower())  # Converts to lowercase: "python"
print(word.capitalize())  # Capitalizes the first letter: "Python"
 ```

####List Methods

Lists have methods that help us add, remove, and sort elements. We haven't dealt with lists that much but we were introduced to it when looking at the range() built-in function and explanations of loops. Lists are simply that, lists of items, those items can be numbers, letters, words, or phrases. Here are some examples of methods for Lists:
```
 numbers = [3, 1, 4, 1, 5]
 numbers.append(9)  # Adds 9 to the end of the list
 numbers.sort()  # Sorts the list in ascending order
 print(numbers)  # Output: [1, 1, 3, 4, 5, 9]
```

####Input Methods

Methods can also be used to modify user input.
```
name = input("What is your name? ").strip()  # strip() removes extra spaces
print("Hello, " + name + "!")
 ```

**Key Takeaways**

* Methods are functions that belong to an object.

* You use a method by typing the object name, followed by a dot (.), and then the method name with parentheses: object.method().

* Different data types (strings, lists, numbers) have their own built-in methods to perform common tasks easily.


These are just some of the methods built in to Python. Like the built-in functions and methods you can also build your own but that will require information about Object Oriented Programming and Classes which we will not cover in this course.

## Conditional Logic
**Comparison Operators (Used for Comparing Values)**

Comparison operators compare two values and return True or False.

Screenshot 2026-05-01 at 2.57.49 PM.png

**Example Code:**
```
a = 5  
b = 10  

print(a == b)  # False  
print(a != b)  # True  
print(a > b)   # False  
print(a < b)   # True  
 ```

**Why Is This Important?**

Comparison operators are useful when making decisions in your code (like in if statements).

`````python
a = 5
b = 10

print(a == b)  # False
print(a != b)  # True
print(a > b)   # False
print(a < b)   # True
`````

**Logical Operators (Used for Making Decisions)**

Logical operators combine multiple conditions to return True or False.
```
x = 5  
y = 10  

print(x > 3 and y > 5)  # True  
print(x > 3 or y < 5)   # True  
print(not(x > 3))       # False  
```
**Why Is This Important?**

Logical operators help when coding decision-making structures like if statements.



`````python
x = 5
y = 10

print(x > 3 and y > 5)  # True
print(x > 3 or y < 5)   # True
print(not(x > 3))       # False
`````

**Assignment Operators (Used for Assigning Values)**

These operators assign values to variables.

Screenshot 2026-05-01 at 3.04.12 PM.png


**Example Code:**
```
x = 10  
x += 5  # Same as x = x + 5  
print(x)  # 15  
 ```

**Why Is This Important?**

Assignment operators help us update variables efficiently.

`````python
x = 10
x += 5  # Same as x = x + 5
print(x)  # 15
`````

**Expressions in Python**

 An expression is a combination of variables, values, and operators that produces a result.



**Examples of Expressions:**
```
result = (10 + 2) * 3  # 36

is_greater = 15 > 10    # True

status = (5 > 3) and (8 < 10)  # True
```
Every time Python evaluates an expression, it returns a value.

**Test It Out:** Try creating your own expressions in a Jupyter Notebook!
