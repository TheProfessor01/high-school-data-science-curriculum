## Part 4: Create a Function

Now we will organize our logic into a function.

Using a function helps us:

- avoid repeating code

- make our program easier to read

- reuse the same logic for every student

### Task

Create a function called `evaluate_student`.

The function should return:

- `"On Track"` if the student has a grade of at least 70 and attendance of at least 80

- `"At Risk"` otherwise



```
def evaluate_student(student):

    if student["grade"] >= 70 and student["attendance"] >= 80:

        return "On Track"

    else:

        return "At Risk"

```
